## Data Acquisition

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
print("Libraries loaded. Pipeline: acquire -> clean -> explore -> analyse -> visualise -> publish.")
```

```python
# Saving raw dataset to CSV and exploring data

air_quality_data.to_csv("air_quality_data.csv", index=False)
AQ_data = pd.read_csv("air_quality_data.csv")
print("Saved and reloaded air_quality_data.csv:", AQ_data.shape)
AQ_data.head(5)
```

## Analyzing the data

```python
# Creating a new column labelled "Region" for better analysis

Region_map = {
    "J&K": "North",
    "Punjab": "North",
    "Haryana": "North",
    "Delhi": "North",
    "Himachal Pradesh": "North",
    "Uttar Pradesh": "North",
    "Madhya Pradesh": "North",   # central, but grouped with North here — see note below

    "Karnataka": "South",
    "Kerela": "South",
    "Tamil Nadu": "South",

    "Bihar": "East",
    "Odisha": "East",
    "West Bengal": "East",

    "Rajasthan": "West",
    "Maharashtra": "West",
    "Gujarat": "West",


    "Sikkim": "Northeast",
    "Arunachal": "Northeast",
    "Assam": "Northeast",
    "Meghalaya": "Northeast"
}

AQ_data["Region"] = AQ_data["State"].map(Region_map)
print("States clubbed under regions:")
AQ_data.head(5)
```

```python
#Ranking top 5 states by average AQI

print("Top 5 states by average AQI:")
print(AQ_data.sort_values("Avg_AQI", ascending=False).reset_index(drop=True)[["State", "Avg_AQI"]].head())
```

```python
# Calculating average regional AQI
print("Average AQI by region (%):")
print(AQ_data.groupby("Region")["Avg_AQI"].mean().round(1).sort_values(ascending=False))
```

```python
#Calculating average good and bad air days by region
print("Average good air days by region (%):")
print(AQ_data.groupby("Region")["Good_air_days"].mean().round(1).sort_values(ascending=False))
print("=========")
print("Average bad air days by region (%):")
print(AQ_data.groupby("Region")["Poor_air_days"].mean().round(1).sort_values(ascending=False))
```

```python
#Calculating correlation betweeen industrial units, green cover percent and avg AQI

print("\nCorrelation with Average AQI:")
correlation = AQ_data[["Avg_AQI", "Industrial_units", "Green_cover_percent"]].corr()
print(correlation["Avg_AQI"].round(2))
```

Correlation analysis
1. Increase in industrial units has strong positive correlation with Avg AQI.
2. Increase in green cover has strong negative correlation with Avg AQI.

```python
#Calculating correlation betweeen pollutants and avg AQI

pollutants = ["Nitrogen_dioxide", "Sulphur_dioxide", "PM_2.5", "PM_10", "Ground_ozone"]
print("\nCorrelation with Average AQI:")
print(air_quality_data[pollutants + ["Avg_AQI"]].corr()["Avg_AQI"].round(2))
```

## Data Visualisation

```python
## Box plot: Distribution of Average AQI Across States and checking outliers

plt.figure(figsize=(6,4))
plt.boxplot(AQ_data["Avg_AQI"], vert=True)
plt.title("Distribution of Average AQI Across States")
plt.ylabel("Average AQI")
plt.tight_layout()
plt.savefig("outputs/aqi_boxplot.png")
plt.show()
```

Takeaway:
The Box plot shows average AQI level across the country hovers between 90 to 140.


```python
## Bar chart: Average AQI by Region

plt.figure(figsize=(7,5))
(AQ_data.groupby("Region")["Avg_AQI"].mean().round(1).sort_values(ascending=False)).plot(kind="bar", color="darkblue")
plt.title("Average AQI by Region")
plt.ylabel("Average AQI")
plt.xlabel("Region")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("outputs/regional_aqi_bar.png")
plt.show()
```

Takeaway:
The Bar chart shows regional rankings in terms of Average AQI with Northern region having maximum average AQI.

```python
## Correlation heatmap: Average Pollutant Levels by Region

pollutants = ["Nitrogen_dioxide", "Sulphur_dioxide", "PM_2.5", "PM_10", "Ground_ozone"]  # use your exact column names
region_pollutant_avg = AQ_data.groupby("Region")[pollutants].mean()
plt.figure(figsize=(8,5))
sns.heatmap(region_pollutant_avg.round(1), annot=True, cmap="YlOrRd")
plt.title("Average Pollutant Levels by Region")
plt.tight_layout()
plt.savefig("outputs/pollutants_by_region_heatmap.png")
plt.show()
```

Takeaway: Region and pollutant correlation
1. Particulate Matter(PM) shows higher concentration levels than rest of the pollutants.
2. PM 10 concentration is highest in the West and North.
2. Ground ozone is of similar value across regions.

```python
## Correlation heatmap: Correlation Between Pollutants and AQI

pollutants = ["Nitrogen_dioxide", "Sulphur_dioxide", "PM_2.5", "PM_10", "Ground_ozone"]
corr_matrix = AQ_data[pollutants + ["Avg_AQI"]].corr()
plt.figure(figsize=(5,5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Pollutants and AQI")
plt.tight_layout()
plt.savefig("outputs/pollutant_correlation_heatmap.png")
plt.show()
```

Takeaway: Correlation heatmap between pollutants and AQI
1. Rise in PM 10 and PM 2.5 pollutants shows strongest correlation to avg AQI.
2. Nitrogen dioxide has a fairly strong correlation with avg AQI.
3. Sulphur dioxide and ground ozone have moderate correlation with avg AQI.

```python
## Grouped bar chart: Average Good vs Poor Air Days by Region

Region_air_days = AQ_data.groupby("Region")[["Good_air_days", "Poor_air_days"]].mean()
Region_air_days.plot(kind="bar", figsize=(9,6), color=["lightgreen", "red"])
plt.title("Average Good vs Poor Air Days by Region")
plt.ylabel("Number of Days")
plt.xlabel("Region")
plt.xticks(rotation=0)
plt.legend(title="Air Quality")
plt.tight_layout()
plt.savefig("outputs/good_vs_poor_air_days_region.png")
plt.show()
```

## Storytelling Panel with insights

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))


# 1. Average AQI by Region
regional_aqi = AQ_data.groupby("Region")["Avg_AQI"].mean().round(1)

regional_aqi.plot(kind="bar", ax=axes[0, 0], color="steelblue")

axes[0, 0].set_title("Average AQI by Region")
axes[0, 0].set_ylabel("Average AQI")
axes[0, 0].set_xlabel("Region")
axes[0, 0].tick_params(axis="x", rotation=0)


# 2. PM2.5 and PM10 by Region
pm_data = AQ_data.groupby("Region")[["PM_2.5", "PM_10"]].mean().round(1)

pm_data.plot(kind="bar", ax=axes[0, 1], color=["orange", "red"])

axes[0, 1].set_title("Average PM2.5 and PM10 by Region")
axes[0, 1].set_ylabel("Average Concentration")
axes[0, 1].set_xlabel("Region")
axes[0, 1].tick_params(axis="x", rotation=0)


# 3. Green Cover by Region
green_cover = AQ_data.groupby("Region")["Green_cover_percent"].mean().round(1)

green_cover.plot(kind="bar", ax=axes[1, 0], color="forestgreen")

axes[1, 0].set_title("Average Green Cover by Region")
axes[1, 0].set_ylabel("Green Cover (%)")
axes[1, 0].set_xlabel("Region")
axes[1, 0].tick_params(axis="x", rotation=0)


# 4. Good vs Poor Air Days
Region_air_days.plot(kind="bar", ax=axes[1, 1], color=["lightgreen", "firebrick"])

axes[1, 1].set_title("Average Good vs Poor Air Days by Region")
axes[1, 1].set_ylabel("Number of Days")
axes[1, 1].set_xlabel("Region")
axes[1, 1].tick_params(axis="x", rotation=0)


plt.suptitle("Regional Air Quality Analysis", fontsize=16)

plt.tight_layout()

plt.savefig("outputs/storytelling_panel.png")

plt.show()
```

## Insights
1. Air quality varies considerably across regions, with some regions recording higher average AQI than others.
2. Higher-AQI regions also show relatively higher PM2.5 and PM10 levels, highlighting particulate matter as an important component of regional air-quality differences.
3. Green cover differs across regions, providing an environmental context for understanding these differences in air quality.
4. The regional differences are also reflected in the number of good and poor air-quality days

Analysis:
1. Regions with high PM levels and low green cover, like North and West, record poor AQI levels in states and lesser number of good air days.
2. Regions with low PM levels and high green cover, like South and NorthEast have recorded better AQI and more number of good air days.
